### Write a Python program using OpenCV to perform histogram equalization on the given grayscale image **"parrot.jpg"**.

The program should:

- Import the required libraries (**OpenCV, NumPy, Matplotlib**).
- Read the image **"parrot.jpg"** in **grayscale** format.
- Display the grayscale image and plot its histogram.
- Perform **histogram equalization** on the grayscale image using `cv2.equalizeHist()` to enhance its contrast.
- Using Matplotlib, display the following in a **2 × 2 grid layout**:
  - Original Grayscale Image
  - Histogram of Original Image
  - Enhanced (Equalized) Image
  - Histogram of Enhanced Image


```python
# Import libraries
import cv2
import numpy as np

import matplotlib.pyplot as plt
```

```python
# Read the image in grayscale format
img = cv2.imread("deerimg.jpg", cv2.IMREAD_GRAYSCALE)

# Check if image is loaded
if img is None:
    raise FileNotFoundError("deerimg.jpg not found")
```

```python
# Display the grayscale image.
plt.imshow(img, cmap="gray")
plt.title("GrayIMAGE")
plt.axis("off")
plt.show()

```

```python
# Plot the histogram of the grayscale image
plt.hist(img.ravel(),256,range=[0, 256])
plt.title("Orginal image")
plt.show()
```

```python
# Perform histogram equalization
equalized_img = cv2.equalizeHist(img)
```

```python
# Display [1] the Original Image (Gray Image) and its Histogram,
# and [2] the Enhanced Image and its Histogram using a 2×2 layout

plt.figure(figsize=(12, 8))

# Original image
plt.subplot(2, 2, 1)
plt.imshow(img, cmap="gray")
plt.title("Original Grayscale Image")
plt.axis("off")

# Original histogram
plt.subplot(2, 2, 2)
plt.hist(img.ravel(), 256, [0, 256])
plt.title("Histogram of Original Image")

# Equalized image
plt.subplot(2, 2, 3)
plt.imshow(equalized_img, cmap="gray")
plt.title("Equalized Image")
plt.axis("off")

# Equalized histogram
plt.subplot(2, 2, 4)
plt.hist(equalized_img.ravel(), 256, [0, 256])
plt.title("Histogram of Equalized Image")
plt.show()
```

### Write a Python program using OpenCV to perform color image enhancement through histogram equalization in the HSV color space on the given image **"parrot.jpg"**.

The program should:

- Import the required libraries (**OpenCV, NumPy, Matplotlib**).
- Read the color image **"parrot.jpg"**.
- Plot the **histogram of the original color image** (use the **B, G, and R** channels).
- Convert the image from **BGR to HSV** color space.
- Apply **histogram equalization** to one specific channel to improve brightness and contrast using `cv2.equalizeHist()`.
- Convert the enhanced HSV image back to **BGR** format.
- Using Matplotlib, display the following in a **2 × 2 grid layout**:
  - Original Color Image
  - Histogram of Original Image (BGR channels)
  - Enhanced Color Image
  - Histogram of Enhanced Image (BGR channels)


```python
# Import required libraries
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Read the color given parrot.jpg image
color_img = cv2.imread("deerimg.jpg")

```

```python
# Plot the histogram of colour image

colors = ('b', 'g', 'r')

plt.figure(figsize=(8, 5))

for i, color in enumerate(colors):
    hist = cv2.calcHist([color_img], [i], None, [256], [0, 256])
    plt.plot(hist, color=color)

plt.title("Histogram of Original Color Image")
plt.xlabel("Pixel Intensity")
plt.ylabel("Frequency")
plt.show()
```

```python
# Convert to HSV
hsv_img = cv2.cvtColor(color_img, cv2.COLOR_BGR2HSV)
```

```python
# Perform histogram equalization

# Split HSV channels
h, s, v = cv2.split(hsv_img)
```

```python
# Equalize the V channel
v_eq = cv2.equalizeHist(v)
```

```python
# Merge channels back
hsv_eq = cv2.merge((h, s, v_eq))
```

```python
# Convert back to BGR format
enhanced_img = cv2.cvtColor(hsv_eq, cv2.COLOR_HSV2BGR)
```

```python
# Display

# Convert BGR to RGB for matplotlib display
original_rgb = cv2.cvtColor(color_img, cv2.COLOR_BGR2RGB)
enhanced_rgb = cv2.cvtColor(enhanced_img, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(12, 6))

# Original image
plt.subplot(1, 2, 1)
plt.imshow(original_rgb)
plt.title("Original Color Image")
plt.axis("off")

# Enhanced image
plt.subplot(1, 2, 2)
plt.imshow(enhanced_rgb)
plt.title("Enhanced Color Image")
plt.axis("off")

plt.tight_layout()
plt.show()
```
